# Visualization of benchmark models

### Sabater1d

```python
from models.library import Sabater1d
import matplotlib.pyplot as plt

# Classical model
model = Sabater1d(design_uq=False)
x_star, q_star = model.load_opt(measure = "quantile", tau = 0.95)
plt_data = model.load_domain_plot(measure = "quantile", tau = 0.95)

plt.plot(plt_data["x"].nodes, plt_data["J"])
plt.plot(x_star, q_star, marker = "+", color = "r", markersize = 10)
plt.title("Classical Sabater 1d model")
plt.xlabel("x")
plt.ylabel("$q_{0.95}$")
plt.show()
```

### Sabater2d

Plot design domain for different stochastic domains

```python
from models.library import Sabater2d
from utils.plotting import contour_plot

uq_domain = "linear"           # 'linear' / 'quad' / 'nonlinear'

model = Sabater2d(uq_domain=uq_domain)
plt_data = model.load_domain_plot(measure = "quantile", tau = 0.95)
x_star, q_star = model.load_opt(measure = "quantile", tau = 0.95)

# Plot
ax = contour_plot(plt_data["x"], plt_data["q"])
ax.plot(x_star[0], x_star[1], marker = "+", color = "r", markersize = 10)
ax.set_xlabel("$x_1$")
ax.set_ylabel("$x_2$")
```

Plot design domain for different robustness measures

```python
from models.library import Sabater2d
from utils.plotting import contour_plot

alpha = 1.0           # 0.0 / 0.5 / 1.0

model = Sabater2d(uq_domain="quad")
plt_data = model.load_domain_plot(measure = "mean_std", alpha = alpha)
x_star, q_star = model.load_opt(measure = "mean_std", alpha = alpha)

# Plot
ax = contour_plot(plt_data["x"], plt_data["J"])
ax.plot(x_star[0], x_star[1], marker = "+", color = "r", markersize = 10)
ax.set_xlabel("$x_1$")
ax.set_ylabel("$x_2$")
```

### Hartmann 6 test function

Plot OAT plot at the optimum

```python
# OAT plot
from models.library import Hartmann6

model = Hartmann6()
oat_plt = model.load_oat_plot(measure = "quantile", alpha = None, tau = 0.95, data_file = None)

import matplotlib.pyplot as plt
fig, ax = plt.subplots(2,3)
ax = ax.flatten()

for i in range(model.n_x):
    ax[i].plot(oat_plt["J_sweeps"][i])
    ax[i].set_xlabel(f"$x_{i+1}$")
    ax[i].set_ylabel("$q_{0.95}$")
    ax[i].grid(True, alpha = 0.3)


fig.tight_layout()
```
