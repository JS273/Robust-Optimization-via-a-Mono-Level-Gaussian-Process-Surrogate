# Finite element bracket model 

This model has been adapted from Archbold, T.A., Kazlauskaite, I., Cirak, F., 2024. Variational bayesian surrogate modelling with application to robust design optimisation. Computer Methods in Applied Mechanics and Engineering 432, 117423.

```python
from models.bracket import create_mesh, fem_simulation, compute_stress, plot_field_solution, plot_von_mises_matplotlib
import numpy as np 
import torch 

# Fixed model parameters
p1 = 50         # right outer edge
p2 = 250        # bottom outer edge
p3 = 250        # left outer edge
p4 = 50         # top outer edge
p5 = 10         # radius corner
p6 = 25         # x coord hole
p7 = 25         # y coord hole
p8 = 5          # Thickness , how to implement this
p = torch.tensor([p1, p2, p3, p4, p5, p6, p7, p8])

x = torch.tensor([60, 30, 30, 60, 30]) 
xi = torch.tensor([5000, np.pi/12])
h = 3 # mesh size parameter
offset = np.array([0.,0.,0.])
circle_radius = 10.0
msh_data = create_mesh(x=x.tolist(), p=p.tolist(), circle_radius=circle_radius, h=h, mesh_save_path= None, offset=offset)
uh, domain = fem_simulation(msh_data,t=p8,f_mag=xi[0],f_angle=xi[1],circle_radius=circle_radius)

# If Xvfb is available --> Plot the displacement field
# plot_field_solution(uh, scale=100)

sig, sig_v = compute_stress(domain, uh)
fig1, ax1, tpc1, data = plot_von_mises_matplotlib(sig_v.x.array.real, domain)
cbar = fig1.colorbar(tpc1, ax=ax1)

```
